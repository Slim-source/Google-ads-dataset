```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
path=r"/content/archive (19).zip"
data=pd.read_csv(path)
print(data.shape)

print(data.head())
```

```python
print(data.isnull().sum())


print(data.dtypes)
```

```python
# print(data['Conversions'])

conversion_mode=data['Conversions'].mode()[0]
data['Conversions']=data['Conversions'].fillna(conversion_mode)


float_cols=data.select_dtypes(include=['float64']).columns
data[float_cols]=data[float_cols].fillna(data[float_cols].mean())


print(data.isnull().sum())
```

```python
data['Cost'] = (
    data['Cost']
    .astype(str)
    .str.replace(r'[\$, ]', '', regex=True)
)

data['Cost']=pd.to_numeric(data['Cost'], errors='coerce')

data['Cost']=data['Cost'].fillna(data['Cost'].mean())


data['Sale_Amount'] = (
    data['Sale_Amount']
    .astype(str)
    .str.replace(r'[\$, ]', '', regex=True)
)

data['Sale_Amount']=pd.to_numeric(data['Sale_Amount'], errors='coerce')

data['Sale_Amount']=data['Sale_Amount'].fillna(data['Sale_Amount'].mean())



print(data.isnull().sum())
```
